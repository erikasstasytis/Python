# Assignment 1: Basic EDA

1. Read in the "Computers.csv" file. 
2. Explore the target variable, "price". Look at a histogram and boxplot of it.
3. Then, produce histograms of the 'speed' and 'ram' variables. 
4. Feel free to further explore the dataset!

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

computers = pd.read_csv("Files/Computers.csv")

computers.head()
```

```python
plt.figure(figsize=(8,5))
plt.hist(computers["price"], bins=30)
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Distribution of Computer Prices")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(computers["price"], vert=False)
plt.xlabel("Price")
plt.title("Boxplot of Computer Prices")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(computers["speed"], bins=20)
plt.xlabel("CPU Speed")
plt.ylabel("Frequency")
plt.title("Distribution of CPU Speed")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(computers["ram"], bins=20)
plt.xlabel("RAM (MB)")
plt.ylabel("Frequency")
plt.title("Distribution of RAM")
plt.show()
```

## Assignment 2: Variable Relationships

1. Build a correlation matrix of numeric features and pass it into a heatmap
2. Build a pairplot of numeric features in the data. Hint: df.select_dtypes("number") selects numeric columns. Consider sampling with df.sample(1000) to speed this up if it takes too long.
3. Build an lmplot of ram vs. price. Does being a premium model (premium == 'yes') impact this relationship? 

```python
numeric_df = computers.select_dtypes("number")

corr = numeric_df.corr()

plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Matrix of Numeric Features")
plt.show()
```

```python
sample_df = numeric_df.sample(500, random_state=42)

sns.pairplot(sample_df)
plt.show()
```

```python
sns.lmplot(
    data=computers,
    x="ram",
    y="price",
    hue="premium",
    height=6,
    aspect=1.2
)
plt.title("RAM vs Price by Premium Status")
plt.show()
```
