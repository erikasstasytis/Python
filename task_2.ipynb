1. Sukurkite sąrašą su skaičiais nuo 1 iki 10. Panaudokite `for` ciklą ir atspausdinkite tik
lyginius skaičius. Naudokite `continue`.

```python
numbers = list(range(1,11))

for n in numbers:
    if n % 2 != 0:
        print(f"{n} is even")
    continue
print(n)
```

2. Turite sąrašą: ['apple', 'banana', 'cherry', 'date']. Naudokite `for` ciklą ir `if`, kad
atspausdintumėte vaisius, kurių ilgis didesnis nei 5.

```python
list = ["apple", "banana", "cherry", "date"]

for fruit in list:
    if len(fruit) > 5:
        print(fruit)
```

3. Naudodami `while` ciklą suskaičiuokite sumą skaičių nuo 1 iki 20. Jei suma viršija 100, ciklą
nutraukite su `break`.

```python
total = 0
i = 1
while i <= 20:
    total += i
    if total > 100:
        break
    i += 1
print(total)
```

4. Turite žodyną su studentų pažymiais. Atspausdinkite tik tuos studentus, kurių pažymys
didesnis arba lygus 8. Naudokite `if` ir `for`.

```python
grades = {'Alice': 8, 'Bob': 9, 'Charlie': 7, 'David': 9}
for student, grade in grades.items():
    if grade >= 8:
        print(student, grade)
```

7. Naudodami `for` ciklą padvigubinkite kiekvieną sąrašo reikšmę: [2, 4, 6]. Rezultatas turi būti
naujas sąrašas.

```python
numbers = [2, 4, 6]
doubled = []

for n in numbers:
    doubled.append(n * 2)
    
print(doubled)
```

8. Iš sąrašo pašalinkite visas reikšmes, mažesnes nei 5. Naudokite `remove()` cikle.

```python
numbers = [1, 3, 5, 8, 10, 12]

for n in numbers[:]:
    if n < 5:
        numbers.remove(n)
        
print(numbers)
```

9. Patikrinkite, ar žodyne yra raktas 'height'. Jei ne, pridėkite jį su reikšme 180.

```python
person = {'name': 'Jonas', 'age': 25}

if 'height' not in person:
    person['height'] = 180

print(person)
```

10. Turite sąrašą [10, 15, 20, 25]. Naudokite `for` ir `if` su `else`, kad atspausdintumėte ar
kiekvienas skaičius didesnis nei 18 ar ne.

```python
numbers = [10, 15, 20, 25]

for n in numbers:
    if n > 18:
        print(f"{n} is greater than 18")
    else:
        print(f"{n} is not greater than 18")
```

11. Naudodami `while` ciklą spausdinkite skaičius nuo 1 iki 5. Jei skaičius lygus 3, praleiskite jį
(`continue`).

```python
i = 1

while i <= 5:
    if i == 3:
        i += 1
        continue
    print(i)
    i += 1
```

12. Turite žodyną su produktų kainomis. Sukurkite naują žodyną su kainomis, padidintomis
10%.

```python
prices = {"apple": 1.2, "banana": 0.5, "cherry": 2.5}

new_prices = {}

for fruit, price in prices.items():
    new_prices[fruit] = price * 1.1  # Increase price by 10%

print(new_prices)
```

13. Sukurkite sąrašą su įvairių tipų duomenimis. Naudokite `for` ir `if`, kad atspausdintumėte
tik `int` tipo reikšmes. Naudokite `type()`.

```python
data = [1, "text" ,3.5, True, None, [1,2,3], {"key": "value"}, 5]

for item in data:
    if type(item) == int:
        print(item)
```
