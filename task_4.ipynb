3. Turite sąrašą kainų [10, 20, 15, 30]. Konvertuokite į set ir parodykite, kiek yra unikalių kainų.

```python
price = [10, 20, 15, 30, 30]
unique_prices = set(price)
print("1) Unikalios kainos:", unique_prices)
print("2) Unikalių kainų kiekis:", len(unique_prices))
```

5. Turite list = ["A", "B", "C"] ir set = {"B", "C", "D"}. Raskite bendrus elementus (naudoti set
operaciją: intersection).

```python
letters_list = ["A", "B", "C"]
letters_set = {"A", "B", "C"}
common = set(letters_list).intersection(letters_set)
print("Bendri elementai:", common)
```

8. Sukurkite list su pirkimo datomis tekstu (pvz. '2023-01-01') ir konvertuokite jas į datetime
objektus naudojant datetime.strptime.


```python
purchase_dates_str = ["2023-01-01", "2024-06-15", "2025-12-31"]
purchase_dates_dt = [datetime.strptime(d, "%Y-%m-%d") for d in purchase_dates_str]
print("8) datetime objektai:", purchase_dates_dt)
```

9. Apskaičiuokite kiek dienų praėjo nuo kiekvienos pirkimo datos iki šiandienos
(datetime.date.today()) ir atspausdinkite.


```python
today = date.today()
for d in purchase_dates_dt:
    days_passed = (today - d.date()).days
    print(f"9) Nuo {d.date()} praėjo {days_passed} dienų")
```

1. Parašykite funkciją, kuri priima skaičių ir grąžina tą skaičių padaugintą iš 2.

```python
def skaicius_kart_du(num):
    return num * 2
print("rezultatas:", skaicius_kart_du(5))
```

2. Sukurkite funkciją, kuri grąžina konversijos rodiklį (conversion rate), jei pateikiamas vartotojų
skaičius ir pirkėjų skaičius. Naudoti: (pirkėjai / vartotojai) * 100.


```python
def conversion_rate(vartotojai, pirkejai):
    return (pirkejai / vartotojai) * 100
```

3. Parašykite funkciją, kuri skaičiuoja CTR (click-through rate): paspaudimai / parodymai * 100.

```python
def ctr(paspaudimai, parodymai):
    return (paspaudimai / parodymai) * 100
```

4. Sukurkite funkciją, kuri apskaičiuoja vidutinę reikšmę iš pateikto sąrašo skaičių. Naudoti
sum() ir len().

```python
def vidurkis(sarasas):
    return sum(sarasas) / len(sarasas)
```

5. Parašykite funkciją, kuri grąžina skirtumą tarp dviejų skaičių – naudinga apskaičiuoti mėnesio
pokytį.

```python
def skirtumas(a, b):
    return a - b
```

6. Sukurkite funkciją, kuri priima pelno ir išlaidų sąrašus ir grąžina bendrą pelną kiekvienam
laikotarpiui (pelno - išlaidos).

```python
def bendras_pelnas(pelnai, islaidos):
    return [p - i for p, i in zip(pelnai, islaidos)]
```

7. Parašykite funkciją, kuri suskaičiuoja, kiek kartų reikšmė 'neaktyvus' yra pateiktame sąraše
naudotojų būsenų.

```python
def skaiciuoti_neaktyvius(busenos):
    return busenos.count("neaktyvus")
```

8. Parašykite funkciją, kuri priima dict su pardavimais pagal regioną ir grąžina regioną su
didžiausia pardavimų suma. Naudoti max().


```python
def geriausias_regionas(pardavimai):
    return max(pardavimai, key=pardavimai.get)
```

9. Sukurkite funkciją, kuri grąžina True, jei pelnas yra teigiamas (didesnis už 0), kitaip – False.

```python
def ar_teigiamas(pelnas):
    return pelnas > 0
```

10. Sukurkite funkciją, kuri gauna list su datomis (tekstais) ir grąžina tų datų sąrašą kaip
datetime objektus. Naudoti datetime.strptime().

```python
from datetime import datetime

def konvertuoti_datas(datos):
    return [datetime.strptime(d, "%Y-%m-%d") for d in datos]
```
