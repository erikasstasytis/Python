<h4># Užduotys su datetime:</h4>

```python
import datetime
```

1. Sukurkite šiandienos datą naudojant datetime.date.today() ir ją atspausdinkite.

10. Patikrinkite, ar duota data yra savaitgalis (šeštadienis arba sekmadienis).

<h4># Užduotys su try/except/finally:</h4>

1. Parašykite kodą, kuris dalija 10 iš 0 ir suvaldo ZeroDivisionError.

```python
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Negalima dalinti iš nulio")
```

2. Bandykite atidaryti neegzistuojantį failą ir suvaldykite FileNotFoundError.

```python
try:
    open("nera.txt")
except FileNotFoundError:
    print("Failas nerastas")
```

3. Pabandykite konvertuoti tekstą 'abc' į int ir suvaldykite ValueError.

```python
try:
    int("abc")
except ValueError:
    print("Neteisingas sveikojo skaičiaus formatas")
```

4. Naudokite try/except/finally – net jei yra klaida, finally turi išspausdinti 'Baigta'.

```python
try:
    10/0
except ZeroDivisionError:
    pass
finally:
    print("Baigta bandyti dalinti")
```

5. Sukurkite kodą, kuriame vyksta dvi galimos klaidos: dalyba iš 0 ir netinkama konversija.

```python
try:
    x = int("xyz")
    print(10/x)
except (ValueError, ZeroDivisionError):
    print("Įvyko klaida")
```

6. Įrašykite skaičių kaip tekstą, konvertuokite į int su try/except.

```python
text = "15"
try:
    number = int(text)
    print(number)
except ValueError:
    print("Negalima konvertuoti į sveikąjį skaičių")
```

7. Sukurkite funkciją, kuri grąžina klaidą, jei įvestas skaičius yra neigiamas.

```python
def check_number(n):
    if n < 0:
        raise ValueError("Skaičius neigiamas")
    return n
print(check_number(-10))
```

8. Naudokite else bloką – jei klaidos nėra, atspausdinkite 'Viskas gerai'.

```python
try:
    x = int("5")
except ValueError:
    print("Klaida")
else:
    print("Viskas gerai")
```

9. Patikrinkite, ar egzistuoja failas 'duomenys.txt' – jei ne, gražiai praneškite.

```python
import os

if os.path.exists("duomenys.txt"):
    print("Failas egzistuoja")
else:
    print("Failo nėra")
```

10. Sukurkite ciklą su try/except, kuris bando padalinti 100 iš sąrašo skaičių, tarp kurių yra ir 0.

```python
numbers = [10, 5, 0, 2]

for n in numbers:
    try:
        print(100 / n)
    except ZeroDivisionError:
        print("Dalyba iš nulio")
```
