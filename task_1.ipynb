8. Apskaičiuokite, kiek procentų liko negrąžintos paskolos (1000 € – 350 € grąžinta).

```python
paskola = 1000
grazinta = 350
likutis = paskola - grazinta
procentai = (likutis / paskola) * 100
print (procentai, "%")
```

21. Patikrinkite, ar skaičius yra teigiamas, neigiamas ar nulis.

24. Įveskite vardą – jei „admin“, pasisveikinkite kaip su administratoriumi.

25. Įveskite temperatūrą: jei >25 – „karšta“, tarp 10–25 – „normali“, kitaip – „šalta“.

```python
temperatura = int(input("Iveskite temperatura Celsijais: "))
if temperatura > 25:
    print("karsta")
elif 10 <= temperatura <= 25:
    print("norma")
else:
    print("salta")        
```
